# Exploratory Data Analysis – Breast Cancer Wisconsin Dataset

## 1. Introduction

This notebook performs Exploratory Data Analysis (EDA) on the Breast Cancer Wisconsin Diagnostic dataset using Python.

The objective is to understand the dataset, examine feature distributions, identify patterns and relationships between variables, and explore differences between malignant and benign tumour samples before developing machine learning classification models.

### Dataset Overview

- **Dataset:** Breast Cancer Wisconsin Diagnostic Dataset
- **Source:** Scikit-learn
- **Total Records:** 569
- **Total Features:** 30 numerical features
- **Target Variable:** Diagnosis
- **Classes:** Malignant and Benign

### Target Variable Mapping

- 0 – Malignant
- 1 – Benign

## 2. Importing Libraries

The following libraries are used for data manipulation, statistical analysis and visualisation.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

sns.set_theme(style="whitegrid")
```

## 3. Loading the Dataset

The dataset is loaded using Scikit-learn's `load_breast_cancer()` function.

The feature matrix is converted into a Pandas DataFrame, and the target variable is added as a separate column.

```python
cancer = load_breast_cancer()

X = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

y = pd.Series(cancer.target, name="target")

df = X.copy()
df["target"] = y
df["diagnosis"] = df["target"].map({
    0: "Malignant",
    1: "Benign"
})

df.head()
```

## 4. Initial Data Inspection

This section examines the structure and quality of the dataset.

The following checks are performed:

- Dataset dimensions
- Column names and data types
- Missing values
- Duplicate records
- Statistical summary

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

### Statistical Summary

The following table provides descriptive statistics for the numerical features, including mean, standard deviation, minimum, maximum and quartiles.

```python
df.describe().T
```

### Observations

- The dataset contains 569 records and 30 numerical input features, along with target information.
- No missing values are present.
- No duplicate records are present.
- The descriptive statistics provide an overview of the range, central tendency and variability of the numerical features.

## 5. Target Variable Distribution

The target variable represents the diagnosis of each tumour sample.

This section examines the distribution of malignant and benign samples to understand whether the dataset is balanced.

```python
class_counts = df["diagnosis"].value_counts()

class_percentages = df["diagnosis"].value_counts(normalize=True) * 100

distribution = pd.DataFrame({
    "Count": class_counts,
    "Percentage": class_percentages
})

print(distribution)
```

### Visualisation

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="diagnosis",
    hue="diagnosis",
    order=["Malignant", "Benign"],
    palette=["#E0679A", "#3FA6BA"],
    legend=False
)

plt.title("Distribution of Tumour Diagnoses")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.show()
```

### Observations

- Malignant samples: 212 (approximately 37.26%).
- Benign samples: 357 (approximately 62.74%).
- The dataset contains more benign samples than malignant samples.
- Although the classes are not equally distributed, both classes have sufficient representation for exploratory analysis.

## 6. Feature Distribution Analysis

This section explores the distributions of selected numerical features using histograms.

The selected features include measurements related to tumour radius, perimeter, area and concavity.

Histograms help understand the spread, frequency and distribution of individual features.

```python
selected_features = [
    "mean radius",
    "mean perimeter",
    "mean area",
    "mean concavity",
    "worst radius",
    "worst concave points"
]

fig, axes = plt.subplots(2, 3, figsize=(15, 9))

for ax, feature in zip(axes.flatten(), selected_features):
    sns.histplot(
        data=df,
        x=feature,
        kde=True,
        ax=ax
    )
    ax.set_title(f"Distribution of {feature}")

plt.tight_layout()
plt.show()
```

### Observations

- The selected features exhibit different distribution patterns.
- Some features show considerable variation across tumour samples.
- The distributions help identify the range and spread of measurements within the dataset.

## 7. Comparative Analysis: Malignant vs Benign

This section compares selected numerical features across malignant and benign tumour samples.

Boxplots are used to examine differences in feature distributions, medians and variability between the two classes.

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

for ax, feature in zip(axes.flatten(), selected_features):
    sns.boxplot(
        data=df,
        x="diagnosis",
        y=feature,
        hue="diagnosis",
        order=["Malignant", "Benign"],
        palette=["#E0679A", "#3FA6BA"],
        legend=False,
        ax=ax
    )

    ax.set_title(feature)
    ax.set_xlabel("Diagnosis")

plt.tight_layout()
plt.show()
```

### Observations

- Several selected features show differences in their distributions between malignant and benign samples.
- Features related to radius, perimeter and area exhibit noticeable differences between the two classes.
- The boxplots also highlight variability and potential outliers in individual measurements.

These patterns suggest that the numerical features may contain useful information for distinguishing between the two diagnosis categories.

## 8. Correlation Analysis

Correlation analysis helps identify relationships between numerical features.

A correlation coefficient indicates the strength and direction of a linear relationship between two variables.

A heatmap is used to visualise the correlation matrix of all 30 numerical features.

```python
correlation_matrix = X.corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.2
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

### Observations

- Several features exhibit strong positive correlations with one another.
- Measurements related to radius, perimeter and area tend to have strong relationships.
- Highly correlated features may contain overlapping information.
- Correlation analysis is useful for understanding relationships between variables before building machine learning models.

Correlation does not imply causation.
